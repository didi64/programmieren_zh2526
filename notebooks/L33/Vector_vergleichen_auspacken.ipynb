### Vektoren vergleichen und auspacken
Wir implementieren einen Vergleichsoperator `==` und Zugriff via Indexnotation
`v[idx]`, plus entsprechende Tests.  

- Die Umwandlung eines Vektors in ein Tuple ist noch etwas umständlich.  
  Wir möchten die Komponenten eines Vektors `v` mit `x, y = v` erhalten.  
  Weiter möchten wir die Komponenten mit `*v` auspacken können,  
  z.B. `canvas.fill_circle(*v, radius=2)`
- Statt 2 Vektoren `u` und `v` mit `u.is_close(v)` zu vergleichen, möchten wir dies mit `u == v` tun können.

Das wird erreicht durch Implementation der folgenden *\_\_dunder\_\_*-Methoden:

```python
def __eq__(self, other):
    ...

def __getitem__(self, idx):
    ...
```

*  `v == w`  triggert nun den Aufruf von `__eq__(v, w)`,
*  `v[i]` trigert den Aufruf von `__getitem__(v, i)`,
*   `x, y = v` übersetzt zu `x, y = v[0], v[1]`.  
*   `f(*v)` wird  zu `f(v[0], ...)`.  
    `v` wird ausgepackt: der Index wird erhöht, bis wird  bis ein `IndexError` auftritt.
* `tuple(v)` liefert `(v[0], ...)`

````python
from vector import Vector as Vec

v = Vec(1, 2)
w = Vec(1, 2)
v == w
````

````python
x, y = v  # v.as_tuple()
````

````python
class Vec:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def is_close(self, other, err=1E-8):
        return abs(other.x - self.x) < err and abs(other.y - self.y) < err

    def __eq__(self, other):
        return self.is_close(other)

    def __getitem__(self, idx):
        if idx == 0:
            return self.x
        elif idx == 1:
            return self.y
        else:
            raise IndexError

    def __repr__(self):
        return f'Vec({self.x:.2f}, {self.y:.2f})'  # Komponenten auf 2 stellen Runden
````

````python
v = Vec(1, 2)
x, y = v
x, y
````

````python
v = Vec(1, 2)
w = Vec(1, 2)
v == w
````
