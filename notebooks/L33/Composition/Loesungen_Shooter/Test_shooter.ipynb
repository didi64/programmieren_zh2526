### Shooter mit Timer

```python
import shooter_mit_timer as shooter
from model_view_controller import BaseView, Controller
from ipycanvas import hold_canvas


class View(BaseView):
    def __init__(self, game):  # erstellt Default-View mit width=height=100 und debug=True
        super().__init__(game)
        self.redraw()

    def redraw(self):
        with hold_canvas(self.canvas):
            self.canvas.clear()
            self.game.cannon.draw(self.canvas)
            s = f'{self.game.timer.displayed_time:06.2f}'  # Zeit als String der Form '012.34'
            self.canvas.fill_text(s, 30, 10)

            for bullet in self.game.bullets:
                bullet.draw(self.canvas)

    def update(self, event, data):
        self.redraw()


def key_handler(self, key, state):
    timer = self.game.timer
    if key == 'Enter':
        if timer.is_running:
            timer.is_running = False
        else:
            timer.start()

    elif key == 'Backspace':
        timer.reset()


game = shooter.Game()

callbacks = {'n': game.new_game,
             ' ': game.shoot,
             'ArrowLeft': game.decrease_angle,
             'ArrowRight': game.increase_angle,
             'ArrowUp': game.step,
             }

game.new_game()
view = View(game)
controller = Controller(game, view, callbacks, key_handler=key_handler)
controller
```
