### Game mit verschiedenen Klassen (und asyncio)

Komposition (engl. Composition) ist ein Designprinzip der objektorientierten Programmierung, 
bei dem eine Klasse Instanzen anderer Klassen als Attribute enthält.

**Beispiel**:  
Bei einem Spiel sollen Bullets mit einer Kanone verschossen werden können.  
Die Klasse Game nutzt dazu Klassen Bullet und Cannon.  
Das Bild zeig 2 schwarze Bullets und die Kanone (roter Strich).

<img src="/files/images/shooter_L33.png">


Ein Bullet-Objekt ist hat eine Position und eine Geschwindigkeit.
Wird die Bullet bewegt, wird der Geschwindigkeitsvektor zu Position addiert.  
Die Kanone soll Bullets verschiessen. 
Die Mündung der Kanone bewegt sich auf einem Kreis um ihre Position.
Die Kanone hat eine Methode `shoot`, die eine Bullet liefert.  

- Die Klasse Bullet hat die Attribute
    ```python
        pos: Vec
        speed: Vec
        radius: float
        health: int  # Lebenszeit der Bullet
        color: str
    ```
    sowie die Methoden `move()` (addiert speed zu pos, dekrementiert health) und `draw(canvas)` (zeichnet die Bullet als Kreis).

- Die Klasse Cannon hat die Attribute
    ```python
        pos: Vec
        radius: float
        angle: float
        delta: float  # um wieviel der Winkel jeweils geaendert wird
        color: str
    ```
    sowie die Methoden `increase_angle`, `decrease_angle` und `shoot()` und `draw(canvas)`.

- Die Klasse Game erbt von Observable und hat die Attribute
    ```python
        cannon: Cannon
        bullets: list[Bullets]
    ```
   sowie die Methoden `new_game`, `shoot`, `increase_angle`, `decrease_angle` und `step()`.

   `step()` aktualisiert die Position der Bullets.
    In einer ersten Version des Spiels rufen wir diese Funktion noch von Hand auf. 

````python
from math import pi as PI
from bullet import Bullet
from cannon import Cannon

# help(Bullet)
# help(Cannon)
````

````python
bullet = Bullet(pos=(0, 0), radius=2, speed=(2, 1), health=5, color='blue')
bullet.move()
bullet
````

````python
cannon = Cannon(pos=(0, 0), radius=10, angle=-PI/2, delta=None)
bullet = cannon.shoot()
bullet
````

````python
from model_view_controller import Observable, notify


class Game(Observable):
    cannon_pos = (50, 50)
    cannon_size = 10

    def __init__(self):
        self.cannon = Cannon(self.cannon_pos, self.cannon_size)
        self.bullets = set()  # Menge von Bullet-Objekten, siehe shoot
        self.is_running = False

    @notify
    def new_game(self):
        self.bullets.clear()

    @notify
    def increase_angle(self):
        self.cannon.increase_angle()

    @notify
    def decrease_angle(self):
        self.cannon.decrease_angle()

    @notify
    def shoot(self):
        self.bullets.add(self.cannon.shoot())

    @notify
    def step(self):
        self._move_bullets()

    def _move_bullets(self):
        for bullet in self.bullets:
            bullet.move()

        self.bullets = {bullet for bullet in self.bullets if bullet.health > 0}

    def __repr__(self):
        return f'Bullets({self.bullets})'
````

````python
game = Game()
game.shoot()
game.increase_angle()
game.shoot()
game
````
