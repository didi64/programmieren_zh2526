### Alternative: Vektorklasse als Unterklasse von Tuple implementieren

Die Klasse `NamedTuple` erlaubt es, die Elemente des Tuples zu benamsen.
Die Elemente können wie Klassen-Attribute mit der dot-Notation angesprochen werden.


```python
from typing import NamedTuple


class Point(NamedTuple):
    x: float  # Name und Typ des 1. Argumentes (wird nicht gecheckt)
    y: float  # Name darf nicht mit _ beginnen!

point = Point(1, 2)
point.x, point.y  # (1, 2)
```


Um aus der Klasse `Point` einen Vektor zu machen, sind nur noch
- die Operation `v + w`, `3*v`, `v*3`, `v / 3`, `v == w` mit den entsprechenden Vektoroperationen zu überschreiben,
- auch noch die Operationen `v - w`, `-v`, `v / 3` zu implementiert.
 
Alles weitere funktioniert wie bei normalen Tuples:
  ```python
     x, y = v
     x, y = v[::-1]
     len(v)  # 2

     for i, component in enumerate(v):
         print(f'{i}-te Komponente: {component}')
  ```

**Achtung**: Bei diese Implementation sind Vektoren immutable. Ihre Komponenten können nicht ge&auml;ndert werden.

````python
from typing import NamedTuple


class Vector(NamedTuple):
    x: float  # wird nicht geprueft (ausser ein Typchecker wie mypy wird genutzt)
    y: float  # int(egers) sind in diesem Sinne auch float(s)

    def __add__(self, other):
        '''gibt den Vektor self+other zurueck'''
        return Vector(self.x + other.x, self.y + other.y)
````

````python
v = Vector(1, 2)
v
````

````python
v.x, v.y  # hat Attribute x und y
````

````python
v + v  # __add__(v, v)
````

````python
v[:]  # Slicenotation wie fuer Tuple
````

````python
for item in v:  # ueber Elemenete iterieren wie bei Tuple
    print(item)
````

````python
v == Vector(1, 2)  # True, falls Elemente gleich sind, wie bei Tuple
````

````python
Vector(1, 2, 3)  # wird Fehler, erwartet 2 Argumente
````

````python
tuple(v)  # Vektor in normales Tuple umwandeln
````
