`get_bad_pairs(numbers)`, welche die Anzahl "schlechter Paare"  zurückgibt.

Ist `numbers` die Liste mit den Zahlen, so ist das Paar
`(numbers[i-1], numbers[i])` gut, falls die Zahl `numbers[i]` rechts von `numbers[i-1]` stehen sollte. Die guten Paare sind (0, 1), (1, 2), ...(18, 19), (19, 0).
Alle andern Paare sind "schlechte Paare". Je weniger "schlechte Paare", desto
näher sind wir bei der Lösung.

```python
def shift_left(numbers):
    return numbers[1:] + numbers[:1]


def shift_right(numbers):
    return numbers[-1:] + numbers[:-1]


def swap4(numbers):
    return numbers[3::-1] + numbers[4:]


def get_neighbors(numbers):
    yield shift_left(numbers)
    yield swap4(numbers)
    yield shift_right(numbers)
```
