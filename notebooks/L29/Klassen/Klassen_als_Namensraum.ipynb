### Klasse als Namensraum

Die einfachste Anwendung einer Klasse ist das Erstellen eines sogenannten Namensraum.  
Alle im Klassenbody einer Klasse `A` definierten Variabeln lassen sich von ausserhalb
der Klasse mit `x.<variabelname>` ansprechen.  

Diese Variabeln können von ausserhalb der Klasse modifiziert werden, und es 
können neue Variabeln definiert werden.

```python
class A:
    x = 42
    text = 'Test'

print(A.x, A.text)  # 42 Test

A.x = A.x + 1
A.y = 1

print(A.x, A.y) # 43 1
```    


In dieser Verwendung verhält sich eine Klasse ähnlich  wie ein Modul:  
- Beim **ersten** Import eines **Moduls** `B` wird der Code des Moduls ausgeführt. 
Nach dem Import ist ein in diesem Modul definiertes **Attribut** `x` (Variable oder Funktion) mit `B.x` ansprechbar.

  Bestehende Attribute können überschrieben und
  neue Attribute können hinzugefügt werden.

- Bei **jedem** Ausführen der **Definition einer Klassen** `A` wird der Code im Klassenbody ausgeführt. Danach ist ein in dieser Klasse definiertes **Attribut** `x` mit
`A.x` ansprechbar. So angesprochene Variabeln und Funktionen verhalten sich
normal.

  Bestehende Attribute können überschrieben und
  neue Attribute können hinzugefügt werden.

**Unterschiede** von Modul und Klasse:
- Modul `B`: Innerhalb einer Funktionsdefinition kann mit `x` eine globale Variable `x` **des Moduls** gelesen werden. 
- Klasse `A`: Innerhalb einer Funktionsdefinition muss mit `A.x` auf das
  Klassenattribut `x` zugegriffen werden. Mit `x` wird auf
  - die lokale Variable `x` der Funktion zugegriffen, **falls definiert**,
  - sonst auf ein globales `x`.

````python
class A:
    x = 42
    text = 'Test'


print(A.x, A.text)  # 42, Test


A.x = A.x + 1
A.y = 1

print(A.x, A.y)
````

````python
x = 0
y = 0


class A:
    x = 42

    def f():
        y = 1
        print(f'A.x: {A.x}')  # x der Klasse
        print(f'x: {x}')      # globales x
        print(f'y: {y}')      # lokales y


A.f()
````

````python
import B

B.f()
B.y = 7
print(B.y)
````

### Beispiel
Bei nachstehender Implementation eines einfachen TikTakToes wird
eine Klasse `State` verwendet, um den Spielzustand zu speichern.  
Folgede Variabeln beschreiben den Spielzustand:
```python
    EMPTY = '.'         # Symbol fuer leeres Feld
    player_chars = 'XO' # Symbol fuer Spieler 0 und Spieler 1
    player_to_move = 0  
    board = EMPTY * 9  # String der Laenge 9, z.B. '.XO.XO.X.'
    result = None      # Resultat, None, 'X', 'O' oder 'draw'
```


Alternativ könnte man auch einen Dict verwenden.
Jedoch ist bei der Klasse der Zugriff mit `State.result` etwas einfacher und hübscher, als beim Dict mit
`State['result']`.

````python
class State:
    '''Namesraum zu Speichern der Variablen,
       die den Spielzustand festlegen
    '''
    EMPTY = '.'
    player_chars = 'XO'
    player_to_move = 0
    board = EMPTY * 9
    result = None


win_lines = [(0, 1, 2), (3, 4, 5), (6, 7, 8),
             (0, 3, 6), (1, 4, 7), (2, 5, 8),
             (0, 4, 8), (2, 4, 6),
             ]


def has_tiktak(player_char):
    for i, j, k in win_lines:
        if State.board[i] == State.board[j] == State.board[k] == player_char:
            return True


def new_game():
    State.board = State.EMPTY * 9
    State.player_to_move = 0
    State.result = None


def play(i):
    if State.result or not (0 <= i < 9) or State.board[i] != State.EMPTY:
        return

    char = State.player_chars[State.player_to_move]
    State.board = State.board[:i] + char + State.board[i+1:]
    if has_tiktak(char):
        State.result = char
    elif State.board.count(State.EMPTY) == 0:
        State.result = 'draw'

    State.player_to_move = 1 - State.player_to_move


def show():
    s = State.board
    print('\n'.join((s[:3], s[3:6], s[6:])))
    print(f'result: {State.result}')
````

````python
for i in (0, 3, 1, 4, 2):
    play(i)
show()
````

````python
new_game()
for i in (0, 3, 1, 4, 6):
    play(i)
show()
````

````python
new_game()
for i in (0, 4, 1, 2, 6, 3, 5, 7, 8):
    play(i)
show()
````
