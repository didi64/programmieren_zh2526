### Spezielle Methoden
Methoden einer Klassen, die mit doppeltem Underscore beginnen und enden
werden auch **Magic- oder dunder-Methods** genannt.  

Die meisten Magic-Methods sind hier erkl&auml;rt,
[www.geeksforgeeks.org/python/dunder-magic-methods-python/](https://www.geeksforgeeks.org/python/dunder-magic-methods-python/).

Eine solche Methode ist `__repr__(self)`.  

- `__repr__(self)`  
    Ist `a` eine Instanz von `A` und wird `a` mit `print(a)` ausgegeben oder ist letzter Ausdruck einer Code-Zelle, wird Etwas der Form `<__main__.A object at 0x7fed0da5f100>` angezeigt.    
    Definiert die Klasse `A` eine Funktion `A.__repr__(self)`, so wird statt dessen
    `A.__repr__(a)` ausgegeben. `A.__repr__(self)` **muss einen String** zurück geben.

- `__add__`
  Sind `x` und `y` Instanzen der Klasse `A`, und trifft der Python-Interpreter auf den Ausdruck `a + b`,
  so wird `A.__add__(a, b)` aufgerufen.

  Ist `A.__add__` nicht definiert, wird ein Fehler erzeugt:  
  `TypeError: unsupported operand type(s) for +: 'A' and 'A'`.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x+other.x, self.y+self.y)

    def __repr__(self):
        return f'Vector({self.x}, {self.y})'


Vector(1, 2)
```

```python
Vector(1, 2) + Vector(3, 3)
```

Schreibe die Klasse `Vector` aus dem  Notebook `Klassen_Intro` um:  
Ersetze `add` und `sub` durch `__add__(self, other)` und `__sub__(self, other)`, und 
füge ` __mul__(self, other)` hinzu, um das Skalarprodukt zu berechnen.

```python
import math


class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def add(self, other):
        '''addiert die Vektoren u und v'''
        return Vector(self.x+other.x, self.y+other.y)

    def sub(self, other):
        '''addiert die Vektoren u und v'''
        return Vector(self.x-other.x, self.y-other.y)

    def rot90(self):
        '''rotiert den Vektor um 90 Grad gegen den Uhrzeigersinn'''
        return Vector(-self.y, self.x)

    def norm(self):
        '''gib die Laenge von v zurueck'''
        return (self.x**2 + self.y**2)**.5

    def mul(self, other):
        '''gib das Skalarproduct von v und w zurueck'''
        return self.x*other.x + self.y*other.y

    def angle(self, other):
        '''gib den Winkel zw. v und w in Radiant zurueck'''
        alpha = math.acos(self.mul(other) / (self.norm()*other.norm()))
        return alpha

    def __repr__(self):
        '''wird automatisch aufgerufen, falls eine
           Stringrepräsentation der Instanz benötigt wird.
           Muss einen String zurück geben.
        '''
        return f'Vec({self.x}, {self.y})'
```
