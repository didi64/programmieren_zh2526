### Sokoban-Levels  
Das Level-Set `levels/Anomaly.txt` ist von 
[https://www.sourcecode.se/sokoban/levels.php](https://www.sourcecode.se/sokoban/levels.php).

Ein einzelnes Level sieht so aus:

```
   #####
####   #
#    # #
# $*.  #
#  *@###
##$*.#
  #   #
  #####
```

- `'#'`: blockiertes Feld
- `'@'`: Spieler
- `'$'`: Box
- `'.'`: Zielfeld (für Box)
- `'*'`: Box auf Zielfeld
- `'+'`: Spieler auf Zielfeld


Ein `level_loader` liest ein Level-File und konvertiert die Levels in eine Liste von 
`NamedTuple`,


```python
class Level(NamedTuple):
    title: str
    player_pos: tuple(int, int)
    boxes: Iterable
    targets: Iterable
    blocked: Iterable
```

````python
from typing import NamedTuple, Iterable
from level_loader import _lines2level, load_levels


class Level(NamedTuple):
    title: str
    ncol: int
    nrow: int
    player_pos: tuple(int, int)
    boxes: Iterable
    targets: Iterable
    blocked: Iterable
````
