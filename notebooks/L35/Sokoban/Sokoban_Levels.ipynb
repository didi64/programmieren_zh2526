### Sokoban-Levels  
Das Level-Set `Levels/Anomaly.txt` ist von 
[https://www.sourcecode.se/sokoban/levels.php](https://www.sourcecode.se/sokoban/levels.php).

Ein einzelnes Level sieht so aus:

```
   #####
####   #
#    # #
# $*.  #
#  *@###
##$*.#
  #   #
  #####
```

- `'#'`: blockiertes Feld
- `'@'`: Spieler
- `'$'`: Box
- `'.'`: Zielfeld (für Box)
- `'*'`: Box auf Zielfeld
- `'+'`: Spieler auf Zielfeld


Die Funktion `load_levels(level_file)` des Modules `level_loader` liest ein Level-File und 
gibt eine eine Liste von `NamedTuple` nachstehender Form zur&uuml;ck.

Die Klasse `LeverHelper` berechnet einige n&uuml;tzliche Dinge eines Levels, die dann
von der Klasse `Solver` benutzt werden.

```python
lh = LevelHelper(level)

lh.cells  # Felder (x, y) des Levels, die der Spieler erreichen kann
lh.G  # Dict, G[pos] ist die Menge der Nachbarn des Feldes pos 

lh.box_cells  # alle Felder, von wo eine Box auf ein Zielfeld gepush werden kann
lh.B_push  # Dict, B_push[box] ist eine Liste mit Elementen (player_pos, box_push).
           # Spieler auf player_pos kann die Box von box nach box_pushed schieben

lh.deadlocks  # Liste mit Mengen von Feldern:
              # Ist auf jedem Feld der Menge eine Box, ensteht ein Deadlock
lh.distances  # distance[(A, B)] ist die Distanz von A nach B.
              # A, B sind Felder in box_cells

lh.get_shortest_path(A, B, blocked=None)  # kuerzerter Weg von A nach B der 
                                          # der Felder in der Menge blocked vermeidet.       


solver = Solver(level)    

solver.solve_bf()  # suche Loesung (Liste mit box_moves)
solver.solve_greedy() # suche Loesung
solver.box_moves2player_walk(box_moves)  # konvertiert box_moves in Player moves
```

````python
from typing import NamedTuple, Iterable


class Level(NamedTuple):
    title: str
    ncol: int
    nrow: int
    player_pos: tuple[int, int]
    boxes: Iterable    # Objekt, ueber das mit for iteriert werden kann,
    targets: Iterable  # wie tuple, list, set, str, generator, ...
    blocked: Iterable
````

````python
from level_loader import _lines2level, load_levels


lines = '''\
  #####
####   #
#    # #
# $*.  #
#  *@###
##$*.#
 #   #
 #####
Title: Level 1\
'''.split('\n')

level = _lines2level(lines)  # lines -> Level
level
````
