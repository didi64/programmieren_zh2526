### Play Sokoban
**Vor** dem Starten eine neuen Spieles mit `n` oder **nach** dem Dr&uuml;cken von `Escape` 
ist man in **levelmode** und man 
mit `ArrowRight` und `ArrowLeft` ein Level w&auml;hlen und dieses mit
BFS (`b`) oder Greedy-Search (`g`) l&ouml;sen und vorspielen lassen.

`n` startet das aktuelle Level, bez. das n&auml;chste Level, falls man das aktuelle gel&ouml;st hat.
Der Spieler wird mit den Pfeiltasten gesteuert.  
`u` (undo) macht die Z&uuml;ge wieder r&uuml;ckg&auml;ngig.


```python
from model_view_controller import Controller
from sokoban import Sokoban, View


def key_handler(self, key, state):
    levelmode = not self.game.started or self.game.done

    if key == 'n':
        self.game.new_game()
    if key == 'u':
        self.game.undo()
    if key == 'Escape':
        self.game.set_levelmode()
    if key == 'b':
        self.game.solve(greedy=False)
    if key == 'g':
        self.game.solve(greedy=True)

    if levelmode:
        if key == 'ArrowLeft':
            self.game.change_level(-1)
        if key == 'ArrowRight':
            self.game.change_level(1)
    else:
        if key == 'ArrowUp':
            self.game.move(0, -1)
        if key == 'ArrowDown':
            self.game.move(0, 1)
        if key == 'ArrowLeft':
            self.game.move(-1, 0)
        if key == 'ArrowRight':
            self.game.move(1, 0)


game = Sokoban()
view = View(game)
controller = Controller(game, view, callbacks=(), key_handler=key_handler)
controller
```
