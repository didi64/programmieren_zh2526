### Grundstrukur eines Spiel mit Modulen

Das Modul `game` ruft immer wenn der Spielzustand ändert eine Funktion `update(event, data)`, die
mitteilt, was passiert ist.

Das Modul `view` überschreibt dann die Funkion `update` des Modules `game`. Die `update`-Funktion der View setzt die Zustandsänderung graphisch um.

De Controller verbindet`game` und `view`.

```python
import game0 as game
import view0 as view
import grid_controller as C


callbacks = {'1': lambda: game.move_to(10, 10),
             '2': lambda: game.move_to(90, 90),
             }

view.init(game)
C.init(view.canvas, callbacks)
C.show(debug=True)
```

###  Das Gleiche Muster mit Klassen

Nun überschreibt ein View-Objekt nicht einfach eine Methode eines Game-Objekts.

Die Game-Klasse erlaubt es, Callbacks zu registrieren, die aufgerufen werden, wwenn sich etwas am Spielzustand &auml;ndert.  
Die Methode `notify` ruft alle registrierten Callbacks auf.

```python
class Game:
    def __init__(self):
        self.player_pos = (50, 50)
        self.callbacks = []

    def register_callback(self, callback):
        self.callbacks.append(callback)

    def notify(self, event, data=None):
        for callback in self.callbacks:
            callback(event, data)

    def move_to(self, x, y):
        self.player_pos = (x, y)
        self.notify('move_to', (x, y))


    def __repr__(self):
        return f'player pos: {self.player_pos}'
```

```python
game = Game()
game
```

```python
import widget_helpers as W


class View:
    def __init__(self, game):
        self.game = game
        self.canvas = W.get_canvas()
        self.canvas.fill_style = 'red'
        self.canvas.fill_circle(*game.player_pos, 5)

        game.register_callback(self.update)


    def update(self, event, data):
        if event == 'move_to':
            x, y = data
            self.canvas.clear()
            self.canvas.fill_circle(x, y, 5)
```

```python
game = Game()
view = View(game)
view.canvas
```

```python
game.move_to(30, 30)
```

```python
game = Game()


callbacks = {'1': lambda: game.move_to(10, 10),
             '2': lambda: game.move_to(90, 90),
             }


view = View(game)
C.init(view.canvas, callbacks)
C.show(debug=True)
```
