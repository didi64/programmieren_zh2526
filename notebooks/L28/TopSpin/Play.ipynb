**Suche**  
- Der Suchraum besteht aus Tuple mit den Zahlen (0, 1, ...,N).  
  Die Nachbarn eines Tupels `s` sind die Tuple `shift_left(s)`, `shift_right(s)` und `swap4(s)`.
  Um zu wissen, welche Operation von einem Zustand zum nächsten führt,
  benutze eine Funktion `get_neighbors(s)`, die Tupels der Form
`(operation, new_state)` liefert. `operation` ist dabei -1 (shift left), 1 (shift right) oder 0 (swap).

- Passe die Suchfunktionen `search_bf`, ... entsprechend an:
  Speichere im Dict `go_back` zu jedem Zustand den alten Zustand und die Operation.  
  `go_back[new_state] = (state, operation)`

- Statt einer Funktion `get_path_home` ist es bequemer, eine Funktion `get_path_to_goal` zu haben.

- Als Heuristik für eine greedy Search bewährt es sich, die Anzahl Zahlen mit einem "falschen" Nachfolger zu berechnen. Zum Beispiel hat das Tuple `(5, 0, 2, 3, 1, 4)`
3 Breakpoints:  0, 3 und 2.
