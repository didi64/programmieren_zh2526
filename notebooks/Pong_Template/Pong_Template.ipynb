```python
from model_view_controller import BaseView
from ipycanvas import hold_canvas


class View(BaseView):
    def __init__(self, game, debug=True):
        super().__init__(game, debug=debug)
        self.redraw()

    def redraw(self):
        with hold_canvas(self.canvas):
            self.canvas.clear()
            self.game.pad.draw(self.canvas)
            self.game.ball.draw(self.canvas)

    def update(self, event, data):
        self.redraw()


game = Pong()
game.new_game()
view = View(game)
view
```

```python
from model_view_controller import Controller

game = Pong()

callbacks = {'ArrowRight': game.step,
             'ArrowDown': game.move_down,
             'ArrowUp': game.move_up,
             }

view = View(game)
controller = Controller(game, view, callbacks)
controller
```
