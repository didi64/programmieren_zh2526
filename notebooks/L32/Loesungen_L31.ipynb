### Vererbung
Die Klassen Circle und Square erben von der Basisklasse Shape.  
Beide Klassen überschreiben die Methoden
```python
   def stroker(self, canvas, anchor, width):
       '''zeichne Umriss des Kreises/Quadrats auf die Leinwand'''
        

    def filler(self, canvas, anchor, width):
        '''zeichnet  Kreis/Quadrat ausgefuellt auf die Leinwand'''
```

````python
class Shape:
    def __init__(self, anchor, width):
        self.anchor = anchor
        self.width = width

    def move(self, dx, dy):
        x, y = self.anchor
        self.anchor = (x+dx, y+dy)

    def stroker(self, canvas, anchor, width):
        print(f'stroke {type(self).__name__} on canvas with center/corner {center} and width/radius {width}')

    def filler(self, canvas, anchor, width):
        print(f'fill {type(self).__name__} on canvas at center/corner {center} and width/radius {width}')

    def draw(self, canvas, clear=True,  fill_style=None, stroke_style='black'):
        canvas.save()

        if clear:
            canvas.clear()
        if fill_style:
            canvas.fill_style = fill_style
            self.filler(canvas, self.anchor, self.width)
        if stroke_style:
            canvas.stroke_style = stroke_style
            self.stroker(canvas, self.anchor, self.width)

        canvas.restore()

    def __repr__(self):
        return f'{type(self).__name__}(anchor={self.anchor}, pts={self.pts})'


class Circle(Shape):
    def __init__(self, center, radius):
        super().__init__(anchor=center, width=radius)

    def stroker(self, canvas, center, radius):
        canvas.stroke_circle(*center, radius)

    def filler(self, canvas, center, radius):
        canvas.fill_circle(*center, radius)

    def __repr__(self):
        return f'Circle(center={self.anchor}, radius=self.width)'


class Square(Shape):
    def __init__(self, lower_left, width):
        super().__init__(anchor=lower_left, width=width)

    def stroker(self, canvas, lower_left, width):
        canvas.stroke_rect(*lower_left, width)

    def filler(self, canvas, lower_left, width):
        canvas.fill_rect(*lower_left, width)

    def __repr__(self):
        return f'Square(lower_left={self.anchor}, width={self.width})'
````

````python
circle = Circle(center=(0, 0), radius=10)
circle.move(50, 50)
square = Square(lower_left=(0, 0), width=20)
square.move(50, 50)
print(circle)
print(square)
````
