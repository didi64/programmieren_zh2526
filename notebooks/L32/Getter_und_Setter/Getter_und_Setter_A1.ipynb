### Aufgaben Getter und Setter

Wir wollen Vektoren nutzen, um ein Polygon bequem verschieben zu können.
Dazu Speichern wir die Punkte und den Ankerpunkt neu als Vektoren. Für den Benutzer soll sich jedoch nichts ändern.

Er soll wie bisher mit `poly = Polygon(pts, anchor)` ein Polygon erstellen, mit
`poly.pts` und `poly.anchor` auf Punkte und Anker zugreifen, und mit
`poly.anchor = new_anchor` den Anker setzen können.

Die Klasse Polygon in der nachstehenden Zelle hat
bereits Getters und Setters `get_pts`, `get_anchor` und `set_anchor`. 


1. Implementiere Setter und Getter mit Hilfe des @property-Decorators.
2. Anschliessend erweitern wir die Klasse Polygon um private Attribute `_size` und `_angle`.
   Implementiere Getter und Setter für diese Attribute.
   - Das Updaten von `size` auf `new_size` soll alle Punkte vom Anker aus um den Faktor
     `new_size / self._size` strecken.
   - Das Updaten von `angle` auf `new_angle` soll alle Punkte um den Winkel
     `new_angle - self._angle` um den Ankerpunkt rotieren.  
     Zeichne dann mit Hilfe folgender Punkte ein F.   
     `pts_F = [(0, 0), (1, 0), (1, 2), (3, 2), (3, 3), (1, 3), (1, 4), (4, 4), (4, 5), (0, 5)]`
  
```python
class Polygon:
    def __init__(self, pts, anchor):
        self._vecs = [Vec(*pt) for pt in pts]
        self._anchor = Vec(*anchor)
        self._size = 1
        self._angle = 0

    def _scale_vecs(self, scale):
        '''scale: Streckungsfaktor
           streckt alle Punkte vom Anker aus
        '''
        c = self._anchor
        self._vecs = [scale*(v-c) + c for v in self._vecs]

    def _rotate_vecs(self, alpha):
        '''rotiere alle Punkte um den Ankerpunkt'''
        c = self._anchor
        self._vecs = [(v-c).rotate(alpha) + c for v in self._vecs]


````python
from vector import Vector as Vec


class Polygon:
    def __init__(self, pts, anchor):
        self._vecs = [Vec(*pt) for pt in pts]
        self._anchor = Vec(*anchor)

    def _translate_vecs(self, dv):
        '''dv: Verschiebungsvektor'''
        self._vecs = [v + dv for v in self._vecs]

    def get_pts(self):
        return [v.as_tuple() for v in self._vecs]

    def get_anchor(self):
        return self._anchor.as_tuple()

    def set_anchor(self, pos):
        new_anchor = Vec(*pos)
        dv = new_anchor - self._anchor  # self._anchor + dv = new_anchor
        self._translate_vecs(dv)
        self._anchor = new_anchor

    def draw(self, canvas, clear=True,  fill_style=None, stroke_style='black'):
        canvas.save()
        if clear:
            canvas.clear()
        if fill_style:
            canvas.fill_style = fill_style
            canvas.fill_polygon(self.get_pts())
        if stroke_style:
            canvas.stroke_style = stroke_style
            canvas.stroke_polygon(self.get_pts())
        canvas.restore()

    def __repr__(self):
        return f'Polygon(anchor={self.get_anchor()}, pts={self.get_pts()})'
````

````python
pts = [(10, 10), (20, 20), (20, 10)]
anchor = pts[0]

triangle = Polygon(pts, anchor=anchor)
triangle
````
