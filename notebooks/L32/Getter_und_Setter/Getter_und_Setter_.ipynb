### Getter und Setter
**Beispiel**:  
Eine Klasse Polygon hat Attribute Punkte (`pts`) und einen Ankerpunkt (`anchor`).  
Wird der Ankerpunkt verschoben, sollen sich alle Punkte mitverschieben.  
Um das zu bewerkstelligen, geben wir der Klasse eine (**Setter**)-Methode `set_anchor`. 


```python
class Polygon:
    def __init__(self, pts, anchor):
        self.pts = pts
        self.anchor = anchor

    def set_anchor(self, new_anchor):
        dx = new_anchor[0] - self.anchor[0]
        dy = new_anchor[1] - self.anchor[1]
        self.pts = [(x+dx, y+dy) for x, y in self.pts]  # Punkte verschieben
        self.anchor = new_anchor                        # Anker setzen
        
    def __repr__(self):
        return f'Polygon(anchor={self.anchor}, pts={self.pts})'
```

Wir möchten, dass ein Benutzer der Klasse `Polygon` das Attribut `anchor` nur mit
der Setter-Methode `set_anchor` modifiziert. Dies wird kommuniziert, indem
das Attribut in `_anchor` umbenannt wird. Ein führender Underscore im Namen signalisiert dem Benutzer der Klasse: Dies ist ein **privates** Attribut, bitte nicht direkt darauf zugreifen!

Damit der Benutzer den Wert des Ankerpunkts lesen kann, schreiben wir eine (**Getter**)-Methode `get_anchor`.  

````python
class Polygon:
    def __init__(self, pts, anchor):
        self.pts = pts
        self._anchor = anchor

    def get_anchor(self):
        return self._anchor

    def set_anchor(self, new_anchor):
        dx = new_anchor[0] - self._anchor[0]
        dy = new_anchor[1] - self._anchor[1]
        self.pts = [(x+dx, y+dy) for x, y in self.pts]  # Punkte verschieben
        self._anchor = new_anchor

    def __repr__(self):
        return f'Polygon(anchor={self.get_anchor()}, pts={self.pts})'
````

````python
center = (50, 50)
pts = [(10, 10), (20, 20), (20, 10)]
triangle = Polygon(pts, pts[0])
triangle
````

````python
triangle.set_anchor(center)
print(f'anchor: {triangle.get_anchor()}')
triangle
````

### Getter und Setter mit dem @property-Decorator

Getter und Setter kontrollieren den Zugriff auf Attribute eines Objekts.
In Python nutzt man für diesen Zweck meistens den @property-Decorator. 
Dies erlaubt, nachträglich festzulegen,
was beim Lesen und Schreiben eines Attributes passiert, ohne den restlichen Code anpassen zu müssen.

```python
class Polygon:
    '''vorher: ohne Getter'''
    def __init__(self, pts, anchor):
        self.pts = pts
        self.anchor = anchor

    def set_anchor(self, new_anchor):
        dx, dy = new_anchor[0] - self.anchor[0], new_anchor[1] - self.anchor[1]
        self.anchor = new_anchor
        self.pts = [(x+dx, y+dy) for x, y in self.pts]

    def __repr__(self):
        return f'Polygon(anchor={self.anchor}, pts={self.pts})'
```


```python
class Polygon:
    '''nachher: mit Getter und Setter'''
    def __init__(self, pts, anchor):
        self.pts = pts
        self._anchor = anchor  # mache anchor zu einem privaten Attribut

    @property  # getter fuer anchor: self.anchor liefert automatisch den Rueckgabewert dieser Methode
    def anchor(self):
        return self._anchor

    @anchor.setter #  setter fuer anchor: self.anchor = position fuehrt diese Methode aus.
    def set_anchor(self, new_anchor):
        dx, dy = new_anchor[0] - self.anchor[0], new_anchor[1] - self._anchor[1]
        self.anchor = new_anchor
        self.pts = [(x+dx, y+dy) for x, y in self.pts]

    def __repr__(self):
        return f'Polygon(anchor={self.anchor}, pts={self.pts})'
```


## Die wichtigsten Gründe, Getter und Setter zu verwenden

* Datenprüfung (Validierung)
* Kapselung: Interne Daten einer Klasse bleiben geschützt. Ändert sich die interne Speicherung, bleibt die Bedienung nach aussen gleich.
* Berechnung bei Bedarf: Ein Getter kann Werte live berechnen, anstatt sie fest im Speicher zu halten.
* Schutz vor Überschreiben: Attribute ohne Setter können nicht unbeabsichtigt überschreiben werden.



````python
class Polygon:
    '''vorher: ohne Getter'''
    def __init__(self, pts, anchor):
        self.pts = pts
        self.anchor = anchor

    def set_anchor(self, new_anchor):
        dx, dy = new_anchor[0] - self.anchor[0], new_anchor[1] - self.anchor[1]
        self.anchor = new_anchor
        self.pts = [(x+dx, y+dy) for x, y in self.pts]

    def __repr__(self):
        return f'Polygon(anchor={self.anchor}, pts={self.pts})'
````

````python
class Polygon:
    '''nachher: mit Getter und Setter'''
    def __init__(self, pts, anchor):
        self.pts = pts
        self._anchor = anchor  # greife nicht direkt auf _anchor zu!

    @property  # getter fuer anchor
    def anchor(self):
        return self._anchor

    @anchor.setter  # setter fuer anchor
    def anchor(self, new_anchor):
        dx, dy = new_anchor[0] - self._anchor[0], new_anchor[1] - self._anchor[1]
        self._anchor = new_anchor
        self.pts = [(x+dx, y+dy) for x, y in self.pts]

    def __repr__(self):
        return f'Polygon(anchor={self.anchor}, pts={self.pts})'
````

````python
center = (50, 50)
pts = [(10, 10), (20, 20), (20, 10)]
triangle = Polygon(pts, pts[0])
triangle.anchor = center  # setter wird automatisch benutzt!
triangle
````
