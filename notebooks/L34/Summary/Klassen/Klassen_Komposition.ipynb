### Komposition von Klassen
Komposition (engl. Composition) ist ein Designprinzip der objektorientierten Programmierung, 
bei dem eine Klasse Instanzen anderer Klassen als Attribute enthält.

**Beispiel**:  
Bei einem Spiel sollen Bullets mit einer Kanone verschossen werden können.  
Die Klasse `Game` nutzt dazu die Klassen `Bullet` und `Cannon`.  

- Die Klasse **Bullet**:  
Ein Bullet-Objekt ist hat eine Position und eine Geschwindigkeit.
Wird die Bullet bewegt, wird der Geschwindigkeitsvektor zu Position addiert. Die Klasse `Bullet` hat die Attribute
    ```python
        pos: Vec
        speed: Vec
        radius: float
        health: int  # Lebenszeit der Bullet
        color: str
    ```
    sowie die Methoden `move()` (addiert speed zu pos, dekrementiert health) und `draw(canvas)` (zeichnet die Bullet als Kreis).

- Die Klasse **Cannon**:  
  Die Mündung der Kanone bewegt sich auf einem Kreis um ihre Position. Die Klasse `Cannon` hat die Attribute
    ```python
        pos: Vec
        radius: float
        angle: float
        delta: float  # um wieviel der Winkel jeweils geaendert wird
        color: str
    ```
    Die Kanone hat eine Methode `shoot`, die eine Bullet liefert,
    sowie die Methoden `increase_angle`, `decrease_angle` und `shoot()` und `draw(canvas)`.

- Die Klasse Game erbt von Observable und hat die Attribute
    ```python
        cannon: Cannon
        bullets: list[Bullets]
    ```
   sowie die Methoden `new_game`, `shoot`, `increase_angle`, `decrease_angle` und `step()`.  
  `step()` aktualisiert die Position der Bullets und wird von einem asynio-Task aufgerufen.

````python
from shooter import Game
from view import View
from model_view_controller import Controller


game = Game()
callbacks = {'n': game.new_game,
             ' ': game.shoot,
             'ArrowLeft': game.decrease_angle,
             'ArrowRight': game.increase_angle,
             'ArrowUp': game.step,
             }

game.new_game()
view = View(game)
controller = Controller(game, view, callbacks)
controller
````
