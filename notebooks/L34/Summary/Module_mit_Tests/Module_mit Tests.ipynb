### Module mit DocStrings und Tests
Module sind `.py`-Files mit Python Code, der beim ersten Import der Moduls ausgeführt wird.
Ein Modul ist wie folgt aufgebaut (vgl. auch `modules/vector.py`):
```python
'''
DocString
Beschreibung des Modules (wird angezeigt mit help(<modulename>)
'''

# Python-Code: 
# Variabeldefinitionen, 
# Funktionen, 
# Klassen
# ...

if __name__ == '__main__':  # wird ausgefuehrt mit %run -m <modulname>
    # Tests zu den Klassen und Funktionen in diesem Module

    def test_fun_1():
        '''teste fun_1 und gib True zurueck 
           falls Test erfolgreich
        '''
        return test_result

    def test_fun_2():
        '''teste fun_2 und gib True zurueck 
           falls Test erfolgreich
        '''
        return test_result

    # Liste mit den Tests    
    tests = [test_fun_1, test_fun_2]
    failed_tests = [test.__name__ for test in tests if not test()]

    n = len(tests)
    m = len(failed_tests)
    print(f'{n-m}/{n} tests passed')
    for test in failed_tests:
        print(f'    Test "{test}" failed')
```


Ein Modul sollte Tests für alle Funktionen/Methoden haben.
- Stellt sicher, dass die Funktionen keine Fehler verursachen und zumindest
  im Sinne der Tests tun, was sie sollten.
- Anwendungsbeispiel für den Benutzer.
- Wird der Code modifiziert, sieht man sofort, ob noch alles wie gehabt funktioniert.
- Wir das Modul importiert, werden die Tests **nicht** angeschaut/ausgeführt.
