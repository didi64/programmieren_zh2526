### Tiefen- und Breitensuche (DFS  und BFS)
**Problem**: Finde einen Weg von einem Startzustand zu einen Zielzustand. Die Nachbarn eines Zustands lassen sich berechnen. Man kann sich den Zustandsraum als Labyrith (Höhlensystem vorstellen).

**Suchstrategien**:
- **DFS**: Unbesuchten Nachbarn werden von rechts an eine Liste `nodes_to_visit` angehängt.
       Jeweils das **letzte** Element von `nodes_to_visit` wird entfernt und besucht.  
       Es wird immer ein neues benachbartes Feld besucht, bis es nicht mehr weiter geht.
       **Irgend ein Weg wird** gefunden.
  
- **BFS**: Unbesuchten Nachbarn werden von rechts an `nodes_to_visit` angehängt.
       Jeweils das **erste** Element von `nodes_to_visit` wird entfernt und besucht.
     
       Auf diese Weise werden zuerst alle Zustände mit Abstand 1 vom Start, dann
       alle Zustände mit Abstand 2 vom Start, u.s.w. besucht. Ein **kürzester Weg** wird gefunden.    
       Nachstehend ist $s_{d,i}$ der $i$-te Zustand mit Abstand $d$ von Start.  
    1. nodes_to_visit = $[start]$ (entferne und besuche start)
    1. nodes_to_visit = $[s_{1,1},\, s_{1,2},\,s_{1,3},\,...,\,s_{1,k}]$ (entferne und besuche $s_{1,1}$)
    1. nodes_to_visit = $[s_{1,2},\,s_{1,3},\,...,\, s_{1,k}] + [s_{2,1}\,, ...,\,s_{2,l}]$ (entferne und besuche $s_{1,2}$)
    1. nodes_to_visit = $[s_{1,3},\,...,\, s_{1,k}] + [s_{2,1}\,, ...,\,s_{2,l}] + [s_{2,l+1}\,, ...,\,s_{2,l'}]$.

**Wann DFS verwenden**?    
Oft kommt man dem Ziel immer einen Schritt näher (8-Damenproblem, Sudoku).
Der Zustand ist eine Teillösungen, ein Nachbarzustand ist eine bessere Teillösungen
(eine Dame mehr platziert, eine Zahl mehr gesetzt.

**Wann BFS verwenden**?              
Nachbarzustand in der Regel nicht näher bei der Lösung.
(Rubiks-Cube, TopSpin).
       
**Varianten**:  
- Greedy Search: Besuche immer den Zustand, der gemäss einer Heuristik am nächsten beim Ziel liegt
  (z.B. TopSpin, wähle Zustand mit den wenigsten *schlechten Paaren*).
- Bidirektionale Search (suche von Start und Ziel her).

### Verwendete Datenstrukturen
- Deque: Eine Liste, bei der Elemente von links und rechts schnell eingefügt und entfernt werden können.
  BFS benutzt aus diesem Grund einen Deque.           
    ```python
    import collections import deque
    
    items = deque([1, 2, 3])
    items.append(4)  # rechts anhaengen
    items.appendleft(0)  # links anhaengen
    print(items)
    items.popleft(), items.pop()  # von links bez. rechts entfernen
    ```
    <br>
    
- Heap: Greedy-Search entfernt jeweils das kleinste Element aus einer Liste `nodes_to_visit`.
        Ein Heap ist eine Datenstruktur, die schnell ein kleinstes Element liefern kann.  
        Die Methoden `headpush(items, item)` und `heappop(items)` des Moduls `heapq` ermöglichen dies:
    - `headpush(items, item)`: nimmt `item` in die Liste `items` auf und ordnet diese geeignet um.
    - `heappop(items)` entfern und liefert ein **kleinsten** Elementes.
    Hat die Liste `items` mehr als ein Element, müssen die Listenelemte zuerst mit
    `heapify(items)` geeignet angeordnet werden.

    ```python
    
    items = [5]  # Liste mit max. einem Element ist schon heapified
    for item in [3, 1, 4]:
        heapq.heappush(items, item)
    
    print(items)
    while items:
        print(heapq.heappop(items))
    ```

**Beispiel Maze generieren und Weg finden**:  
Auf einem leeren $m \times n$ Schachbrett starten wir
auf dem Feld oben links und suchen mit DFS ein besetztes Feld:
- Hänge alle unbesuchten Nachbarn des aktuellen Feldes in zufälliger Reihenfolge an eine Liste
  `nodes_to_visit` an.
- Entferne das letzte Element der Liste `nodes_to_visit` und besuche dieses Feld.

Da es kein besetztes Feld gibt, endet die Suche erfolglos, sobald wir **alle** Felder besucht haben.
Benachbarte Felder `f1` und `f2` betrachen wir als verbunden, falls `f1 in go_back[f2]`.   
Da jedes Feld nur einmal besucht wird, gibt es zu jedem Feld genau einen Weg.
Um das zu ändern, fügen wir noch einige zufällige Verbindungen hinzu.

Mit DFS und BFS suchen wir nun
Wege von links oben nach rechts unten (siehe `maze.py` und `view_maze.py`).  

````python
from maze import Maze
from view_maze import ViewMaze as View
from model_view_controller import Controller


maze = Maze()
callbacks = {'n': maze.new_maze,   # erstellt zufaelliges Maze mit DFS
             'a': maze.add_random_connections,   # fuege Verbindungen hinzu
             'd': lambda: maze.solve(df=True),   # finde Pfad mit DFS
             'b': lambda: maze.solve(df=False),  # finde Pfad mit BFS
             }

view = View(maze)
controller = Controller(maze, view, callbacks)
controller
````
