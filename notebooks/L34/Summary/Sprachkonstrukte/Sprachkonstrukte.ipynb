### Sprachkonstrukte

**Rechnen mit `and` und `or` und Booleans**:  
Ein Ausdruck `expr` heisst *truthy* falls `bool(expr)` True ist, sonst *falsy*.  
Beim Rechnen verhält sich `True` wie `1` und False wie `0`.

`x and y` ist gleich
- `x`, falls `x` *falsy* ist (das 2. Argument `y` wird **nicht** ausgewertet)
- `y`, falls `x` *truthy* ist.
  
`x or y` ist gleich
- `x`, falls `x` *truthy* ist (das 2. Argument `y` wird **nicht** ausgewertet)
- `y`, falls `x` *falsy* ist.
  
```python
def draw_circle(canvas, pt_1, pt_2, center=None):
    '''zeichne einen Kreis durch pt_1 und pt_2 und den 
       Mittelpunkt von pt_1 und pt_2, oder center
    '''
    center = center or (pt_1 + pt_2) / 2  # center, falls center nicht None
    radius = (pt_1 - center).norm()
    canvas.stroke_circle(*center, radius)

def sign(x):
    '''1,  falls x > 0,
       -1, falls x < 0,
       0,  falls x - 0.
    '''
    return (x > 0) - (x < 0)  
```

**Walruss Operator `:=`**:  
`(x := <Ausdruck>)` weist der Variable `x` den Wert des Ausdrucks zu und gibt diesen Wert zurück.  
Hilft, dass Wiederholen von Berechnungen zu vermeiden.  
```python
def find_occurrences(word, char):
    '''liefert Liste mit den Indices, wo char in word vorkommt'''
    idxs = []
    i = 0
    while (i := word.find(char, i)) != -1:
        idxs.append(i)
        i += 1
    return idxs
```

**Generator-Funktion**  
Funktion `f` mit `yield` Anweisung. 
Das Ausführen der Funktion liefert einen Generator `gen = f()`.
- `next(gen)` liefert ein nächstes Element oder einen `StopIterationError`:  
  Die Funktion läuft bis zum nächstes `yield` und gibt den Wert des
  Ausdrucks nach `yield` zurück. Wird das Ende der Funktion oder ein `return` erreicht, wird
  `StopIterationError` ausgelöst.
- Ein Generator kann man mit `list(gen)` zu einer Liste machen, oder mit einem
  For-Loop darüber iterieren.

```python
def get_neighbors(col, row, ncol, nrow):
    '''liefert alle 4-Nachbarn des Feldes (col, row) in einem ncol x nrow Gitter'''
    for dc, dr in ((0, -1), (-1, 0), (1, 0), (0, 1)):
        if (0 <= (col_new := col+dc) < ncol) and (0 <= (row_new := row+dr) < nrow):
            yield col_new, row_new
```

```python
# Beispiele von Decorators
from functools import wraps

def save_canvas_state(f):
    '''save_canvas_state(f) liefert eine Funktion, die
       - canvas.save() ausführt
       - f(canvas, ...) ausführt
       - canvas.restore() ausführt
    '''
    @wraps(f)
    def wrapper(canvas, *args, **kwargs):
        canvas.save()
        result = f(canvas, *args, **kwargs)
        canvas.restore()
        return result
        
    return wrapper
    
    
def notify(f):
    '''f(self, ...) ist eine Methode.
       notify(f) liefert eine Funktion, die  
       - data = f(self, ...) ausführt
       - self._notify(f.__name__, data) aufruft
    '''
    @wraps(f)  # korrigiert u.a. f.__name__
    def wrapper(self, *args, **kwargs):
        data = f(self, *args, **kwargs)
        self._notify(f.__name__, data)
    return wrapper
```

### Decorators

````python
from functools import wraps


def args2str(args, kwargs):
    arg_str = ', '.join(f'{arg!r}' for arg in args)
    kwargs_str = ', '.join(f'{k}={v!r}' for k, v in kwargs.items())
    return f'{arg_str}, {kwargs_str}'


def show_args(f):
    @wraps(f)
    def wrapper(*args, **kwargs):
        print(f'calling {f.__name__}({args2str(args, kwargs)})')
        result = f(*args, **kwargs)
        return result

    return wrapper
````

````python
@show_args
def g(x, y, z):
    return x*y + z
````

````python
res = g(3, y='foo', z='bar')
````

### Fehler auslösen/abfangen:

````python
def divide_1(x, y):
    assert isinstance(x, (int, float)) and isinstance(y, (int, float)), 'x and y must be numbers!'
    if y == 0:
        raise ValueError('y must not be 0!')
    return x / y


def divide_2(x, y):
    try:
        result = x / y
    except TypeError, ZeroDivisionError:
        print(f'x and y must be numbers and y must not be 0, got {x!r} and {y!r}')
        result = None
    return result
````

````python
divide_1(1, 'e')  # AssertionError wird ausgeloest
````

````python
divide_2(1, 'e')  # Fehler wird abgefangen
````
