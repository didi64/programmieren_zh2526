### Die Dekoratoren `classmethod` und `staticmethod`

Sei `A` eine Klasse und `a=A()` eine Instanz von `A`.  
Obige Dekoratoren legen fest, ob beim Aufruf 
`A.f(...)` und `a.f(...)` ein zusätzliches erstes Argument eingefügt wird, 
und ob diese `a` oder `A` ist.


```python
class A:
   @classmethod
   def f(cls, *args):
       '''A.f() und a.f() erhalten die Klasse A 
          als zusätzlicheserstes Argument
       '''
       ...
       
   @staticmethod
   def g(*args):
       '''A.g() und a.g() erhalten KEIN
          zusätzlicheserstes Argument
        '''
       ...
``` 
**Aufgabe**:
Ergänze nachstehende Klasse und erstelle eine Instanz `a = A()`.
Teste für die Methode `f` (und ebenso `g` und `h`), welches Objekt beim Aufruf
via `A.f()` und `a.f()` jeweils an das erste Arument gebunden wird.

```python
class A:
    def f(*args):
        '''gib Laenge und Inhalt des Tuples args aus'''
        ...
        
    @classmethod
    def g(*args):
        '''gib Laenge und Inhalt des Tuples args aus'''
        
    @staticmethod
    def h(*args):
        '''gib Laenge und Inhalt des Tuples args aus'''
```

### Typische Verwendungsweisen von `staticmethod` und `classmethod`



**classmethod**:
Ist eine Methode einer Klasse damit dekoriert,
```python
class A:
    @classmethod
    def f(cls, ...):
        ...
```
so wird das ersten Argument, welches wir `cls` (für `class`) nennen, immer
an die Klasse `A` gebunden. Innerhalb dieser Funktion verwendet man dann stets `cls(...)` für `A(...)`. Verwendet wird dieser Dekorator hauptsächlich in folgender Situation:

- zusätzlicher Konstruktor
```python
class Vector
   @classmethod
   def from_str(cls, s):
       '''erstelle einen Vector aus einen String 
          der Form `s = (2, 3)`
       ''' 
       a, b = s[1:-1].split(',')
       return cls(int(a), int(b))
      

s = '(2, 3)`
Vector.from_str(s)
```

**staticmethod**:
Ist eine Methode einer Klasse damit dekoriert, wird nie ein
zusätzliches erstes Argument eingefügt.
Die Funktion verhält sich, als ob sie ausserhalb der Klasse definiert worden wäre.


- Hilfsfunktionen, unabhängig von der Klasse und Instanz.

```python
class Vector
   @staticmethod
   def rad2deg(alpha):
       return alpha * 180/3.141
```

````python
def rad2deg(alpha):
    return alpha * 180/3.141


class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    @classmethod
    def from_str(cls, s):
        '''erstelle einen Vector aus einen String
           der Form `s = (2, 3)`
        '''
        a, b = s[1:-1].split(',')
        return cls(int(a), int(b))

    def __repr__(self):
        return f'Vec({self.x}, {self.y})'
````

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    @staticmethod
    def rad2deg(alpha):
        return alpha * 180/3.141

    @classmethod
    def from_str(cls, s):
        '''erstelle einen Vector aus einen String
           der Form `s = (2, 3)`
        '''
        a, b = s[1:-1].split(',')
        return cls(int(a), int(b))

    def __repr__(self):
        return f'Vec({self.x}, {self.y})'
````

````python
v = Vector(1, 2)
v
````

````python
Vector.rad2deg(1.57)
````
