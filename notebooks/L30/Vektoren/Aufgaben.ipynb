Benutze dazu die (vervollständigte) Klasse Polygon.  
Die `__init__`-Methode erhält u.a. eine Liste `pts` der Punkte `(x, y)` des Polygons.
Der Ankerpunkt `anchor=(x0, y0)` ist dabei der Punkt, 
von dem aus das Polygon gestreckt wird und
um den rotiert wird.
Wird das Polygon an eine neue Position `pos=(x1, y1)` verschoben, 
so wird der Ankerpunkt (mit allen Punkte des Polygons) auf `pos` geschoben.
Alle Punkte werden also um `Vec(x1, y1) - Vec(x0, y0)` verschoben.



````python
from vector import Vector as Vec
import math
from math import pi as PI


class Polygon:
    def __init__(self,
                 pts,
                 anchor=(0, 0),
                 size=1,    # Polygon wir von anchor aus um diesen Faktor gestreckt
                 alpha=0,   # Polygon um alpha um anchor gedreht
                 fill=True,
                 stroke=True,
                 pos=None,  # anchor wird mit Polygon auf pos geschoben
                 ):
        self.vecs = [Vec(*pt) for pt in pts]
        self.anchor = Vec(*anchor)
        self.size = 1
        self.alpha = 0
        self.fill = fill
        self.stroke = stroke

        self.update_size(size)
        self.update_alpha(alpha)
        if pos:
            self.update_anchor(pos)

    def copy(self):
        '''gib Polygon(...) zurueck'''
        ...

    def update_size(self, size):
        '''strecke jeden Punkt v in self.vecs von self.anchor aus
           mit dem Faktor s = size/self.size, bevor self.size = size gesetzt wird
           Der Ankerpunkt self.anchor soll sich nicht aendern:
           v -> s*(v - self.anchor) + self.anchor
        '''
        ...

    def update_anchor(self, pos):
        '''verschiebe self.anchor und self.vecs um Vec(*pos) - self.anchor,
           bevor self.anchor = pos gesetzt wird
        '''
        ...

    def update_alpha(self, alpha):
        '''rotiere jeden Vektor v in self.vecs um den Winkel phi = alpha - self.alpha,
           bevor self.alpha = alpha gesetzt wird
           Der Ankerpunkt self.anchor soll sich nicht aendern:
           v -> (v - self.anchor).rotate(phi) + self.anchor
        '''
        ...

    def rotate(self, center, alpha):
        '''rotiere jeden Vektor v in self.vecs um alpha um den Punkt center,
           bevor self.alpha += alpha wird.
           Der Ankerpunkt self.anchor soll sich nicht aendern:
           v -> (v - Vec(*center)).rotate(alpha) + Vec(*center)
        '''
        ...

    @staticmethod
    def _set_canvas_attrs(canvas, kwargs):
        for k, v in kwargs.items():
            if k in canvas.ATTRS:
                setattr(canvas, k, v)

    def get_pts(self):
        return [v.as_tuple() for v in self.vecs]

    def draw(self, canvas, clear=True, flip=True, **kwargs):
        canvas.save()

        pts = self.get_pts()
        if flip:
            pts = [(x, canvas.height - y) for x, y in pts]
        self._set_canvas_attrs(canvas, kwargs)

        if clear:
            canvas.clear()
        if self.fill:
            canvas.fill_polygon(pts)
        if self.stroke:
            canvas.stroke_polygon(pts)

        canvas.restore()
````
