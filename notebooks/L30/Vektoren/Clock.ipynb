### Uhr mit Zeiger
Ziel ist es, die aktuelle Uhrzeit zu erfragen, und dann die
Uhrzeiger entsprechend auf das Zifferblatt zu malen.

<img src="/files/images/L30_ticking_clock.png">

```python
from datetime import datetime


def get_time():
    now = datetime.now()
    hours = now.hour
    minutes = now.minute
    seconds = now.second
    return hours, minutes, seconds
```
