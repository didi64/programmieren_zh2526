### Ausbau der Klasse `Vector`
Wir möchten aber auch einen Vektor mit einer Zahl von recht und links multiplizieren können,
`v*3` und `3*v`.

Im Fall `v*3` wird `v.__mul__(3)`, also `Vector.__mul__(v, 3)` aufgerufen. Hier können wir
in dieser Funktion den Typ des 2. Argumentes mit `isinstance` testen:

```python
class Vector:
    def __mul__(self, other):
        if isinstance(other, Vector):
            # berechene Skalarprodukt
        elif isinstance(other, (int, float)):
            # strecke den Vektor
        else:
            return NotImplemented    
```

Im Fall `3*v` wird `int.__mul__(3, v)` aufgerufen, was `NotImplemented` zurückgibt.  
Als nächstes versucht der Python-Interpreter  `v.__rmul__(s)`, also
`Vector.__rmul__(v, s)` aufzurufen. Wenn auch das `NotImplemented` zurückgibt oder nicht existiert, wirft Python einen `TypeError`.

Damit `3*v` den Vektor `v` mit `3` multipliziert, müssen wir also nur
```python
    def __rmul__(self, s):
        return self * s  # ruft Vector.__mul__(self, s) auf,
                         # gibt korrektes Resultat falls s Zahl
```
in der Klasse `Vector` haben.

````python
s = 3
# s = 3.0
isinstance(s, (int, float))  # True, falls s vom Typ int oder float
````

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __mul__(self, other):
        if isinstance(other, Vector):
            return self.x+other.x + self.y+self.y
        elif isinstance(other, (int, float)):
            return Vector(other*self.x, other*self.y)
        else:
            return NotImplemented

    # def __rmul__(self, s):
    #     return self * s

    def __repr__(self):
        return f'Vector({self.x}, {self.y})'
````

````python
v = Vector(1, 2)
v*v, v*3, 3*v
````
