### Anwendungen der Klasse Vector

Wir nutzen die Klasse Vector, um
Polygone zu strecken, verschieben, drehen und spiegeln.  
Als erste Anwendung schreiben wir eine Funktion, die einen Pfeil auf die Leinwand zeichnet.

Als Pfeilspitze verwenden wir das Dreieck `[(-2, 1), (-2, -1), (0, 0)]`.
Um einen Pfeil von `start=(50, 50)` nach `end=(75, 30)` zu zeichnen,
drehen, skalieren und verschieben wir dieses Dreieck so,
dass seine Spitze auf den Endpunkt zeigt. Siehe unten.

```python
import math
from vector import Vector as Vec
```

```python
# arrow: Vektor von start nach end
start = Vec(50, 50)
end = Vec(75, 30)
arrow = end - start
length = arrow.norm()
arrow, length
```

```python
# Dreieckspunkte in Vektoren verwandeln
tip = ((-2, 1), (-2, -1), (0, 0))
tip = [Vec(x, y) for x, y in tip]
tip
```

```python

alpha = math.atan2(arrow.y, arrow.x)
alpha * 180/math.pi
```

```python
rotated_tip = [v.rotate(alpha) for v in tip]
rotated_tip
```

```python
s = length/10
translated_and_scaled_tip = [end + w*s for w in rotated_tip]
translated_and_scaled_tip
```

### Aufgaben
- Schreibe eine Funktion `draw_arrow(canvas, start, end, tip_size=5, color='red')`,
  welche einen Pfeil von `start` nach `end` zeichnet. `tip_size` ist dabei die Höhe des Pfeils.
  `start` und `end` sind Koordinaten-Tuple.
  

```python
from decorators import save_canvas_state


@save_canvas_state
def draw_arrow(canvas, start, end,
               tip_size=5, line_width=1,
               color='black', tip_color='red'):
    start = Vec(*start)
    end = Vec(*end)
    arrow = end - start
    alpha = math.atan2(arrow.y, arrow.x)

    tip = [Vec(*pt) for pt in ((-2, 1), (-2, -1), (0, 0))]
    rotated_tip = [v.rotate(alpha) for v in tip]
    translated_and_scaled_tip = [end + w*tip_size for w in rotated_tip]
    pts = [v.as_tuple() for v in translated_and_scaled_tip]

    canvas.line_width = line_width
    canvas.stroke_style = color
    canvas.fill_style = tip_color
    canvas.stroke_line(*start.as_tuple(), *end.as_tuple())
    canvas.fill_polygon(pts)
```
