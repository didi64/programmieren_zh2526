Um Start- und Endpunkt der $i$ Minuten Markierung zu finden,
starten wir mit einem Vektor  
$v = (\cos(2i\pi/60), \sin(2i\pi/60))$,  
der auf dem Einheitskreis an die richtige Position zeigt.  
Start- und Endpunkt der Markierung sind $0.9\cdot v$ und $1.1\cdot v$.

Ist das Zentrum des Zifferblatts beim Punkt $c=(x_0, y_0)$ und hat Radius $r$,
verschieben sich Start- und Endpunkt der Markierung nach
$0.9r \cdot v + c$ und $1.1r \cdot v + c$.

```python
import math
from math import pi as PI
from vector import Vector as Vec

center = (50, 50)
radius = 40

marks = [(math.cos(i*2*PI/60), math.sin(i*2*PI/60)) for i in range(60)]
starts = Vec.transform(marks, center, scale=0.9*radius)
ends = Vec.transform(marks, center, scale=1.1*radius)

lines = [[start, end] for start, end in zip(starts, ends)]
```

```python
def draw_clockface(canvas, center, radius):
    '''zeichet ein Zifferblatt am Punkt center mit geg. radius'''
    marks = [(math.cos(i*PI/30), math.sin(i*PI/30)) for i in range(60)]
    starts = Vec.transform(marks, center, scale=0.9*radius)
    ends = Vec.transform(marks, center, scale=1.1*radius)

    lines = [[start, end] for start, end in zip(starts, ends)]

    canvas.save()
    for i, line in enumerate(lines):
        if i % 5 == 0:
            canvas.line_width = 3
            canvas.stroke_style = 'blue'
        else:
            canvas.line_width = 1
            canvas.stroke_style = 'black'

        canvas.stroke_lines(line)

    canvas.restore()
```
