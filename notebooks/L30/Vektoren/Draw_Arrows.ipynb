### Pfeile zeichnen

Wir nutzen die Klasse Vector, um
Polygone zu strecken, verschieben, drehen und spiegeln.  
Als erste Anwendung schreiben wir eine Funktion, die einen Pfeil auf die Leinwand zeichnet.

Als Pfeilspitze verwenden wir das Dreieck `[(-2, 1), (-2, -1), (0, 0)]`.
Um einen Pfeil von `A` nach `B` zu zeichnen,
drehen, skalieren und verschieben wir dieses Dreieck so,
dass seine Spitze auf den Endpunkt zeigt.
- Rotationswinkel $\varphi$: Winkel zw. $x$-Achse und dem Vektor BA=w-v.  
  $\varphi$ = `math.atan2(BA.y, BA.x)`  (**Achtung**: y-Koord. ist das 1. Argument!)

```python
import math
from math import pi as PI
from vector import Vector as Vec
```

```python
math.atan2(5, 5) * 180/PI
```

```python
# arrow: Vektor von start nach end
start = (50, 50)
end = (75, 30)
arrow = Vec(*end) - Vec(*start)
length = arrow.norm()
arrow, length
```

```python
# Dreieckspunkte in Vektoren verwandeln
tip_pts = [(-2, 1), (-2, -1), (0, 0)]
tip = [Vec(x, y) for x, y in tip_pts]
tip
```

```python
alpha = math.atan2(arrow.y, arrow.x)
alpha * 180/PI
```

```python
rotated_tip = [v.rotate(alpha) for v in tip]
rotated_tip
```

```python
scale = length/10  # Streckungsfaktor
dv = Vec(*end)
translated_and_scaled_tip = [scale*v + dv for v in rotated_tip]
translated_and_scaled_tip
```

```python
end
```

```python
Vec.transform(tip_pts, end, alpha, scale)
```

```python
import widget_helpers as W


canvas = W.get_canvas()
canvas
```

```python
canvas.stroke_line(*start, *end)
```

```python
canvas.fill_style = 'red'
pts = [v.as_tuple() for v in translated_and_scaled_tip]
canvas.fill_polygon(pts)
```

```python
canvas.stroke_line(*start, *end)
```
