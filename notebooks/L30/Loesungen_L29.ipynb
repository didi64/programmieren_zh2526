```python
class Student:
    def __init__(self, name):
        '''initialisiert eine Instanz der Klasse Student'''
        self.name = name
        self.grades = {}

    def add_grade(self, subject, grade):
        '''Key-Value-Paar (subject, grade) in den Dictionary grades aufnehmen'''
        if subject not in self.grades:
            self.grades[subject] = [grade]
        else:
            self.grades[subject].append(grade)

    def __repr__(self):
        return f'Student({self.name})\nNoten: {self.grades}'
```

```python
bob = Student('Bob')
bob
```

```python
bob.add_grade('Math', 6)
Student.add_grade(bob, 'Math', 5)
bob
```

### Aufgaben
Die Studierenden Alice, Bob und Carl besuchen die Kurse English und Mathe.
Der Dict `noten` in der nächsten Zelle zeigt ihre Noten.

1. Schreibe eine Funktion `init_students(noten)`, die eine Liste mit Objekten vom Typ `Student`
liefert. Das erste Element dieser Liste ist Alice mit ihren Noten, u.s.w.

2. Schreibe eine Funktion `show_marks(students)`, die zu jedem Studenten in der Liste `students`
   den Namen, sowie die besuchten Kurse mit der Durschnittsnote ausgibt.

```python
noten = {
    'Alice': {'Englisch': [4, 5.5, 3.5],
              'Mathe':    [5, 5, 4, 5],
              },
    'Bob':   {'Englisch': [6, 5.5, 2],
              'Mathe':    [4, 4, 4.5],
              },
    'Carl':  {'Englisch':  [4.5, 4.5],
              'Mathe':     [6, 2, 3.5],
              },
}
```

```python
def init_students(noten):
    students = []
    for name, lectures in noten.items():
        student = Student(name)
        for subject, grades in lectures.items():
            for grade in grades:
                student.add_grade(subject, grade)
        students.append(student)

    return students
```

```python
students = init_students(noten)
students
```

```python
def show_marks(students):
    for student in students:
        print(student.name)
        for subject, grades in student.grades.items():
            print(f'    {subject}: Durchschnitt: {sum(grades)/len(grades):.2f}')
        print('-'*40)
```

```python
show_marks(students)
```
