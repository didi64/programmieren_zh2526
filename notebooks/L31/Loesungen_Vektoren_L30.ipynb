### Aufgaben zu Vektoren (L30)

```python
from vector import Vector as Vec
import loesungen_vektoren as L
import widget_helpers as W
from math import pi as PI
from ipywidgets import HBox


canvas1 = W.get_canvas()
canvas2 = W.get_canvas()
canvas3 = W.get_canvas()
HBox([canvas1, canvas2, canvas3])
```

```python
center = (50, 50)
radius = 30
n = 5
colors = ['blue', 'lightblue', 'yellow', 'orange', 'red']

for i in range(n):
    v = radius*Vec.e1.rotate(2*i*PI/n) + Vec(*center)
    end = v.as_tuple()
    L.draw_arrow(canvas1, center, end, line_width=2,
                 color=colors[(i+1) % len(colors)], tip_color=colors[i % len(colors)])
```

```python
center = (50, 50)
radius = 40
hour, minute, second = (9, 42, 17)
L.draw_clockface(canvas2, center, radius)
L.draw_hour_hand(canvas2, center, radius, hour, minute)
L.draw_minute_hand(canvas2, center, radius, minute)
L.draw_second_hand(canvas2, center, radius, second)
```
