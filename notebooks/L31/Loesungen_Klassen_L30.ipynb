**Einfluss von staticmetod und classmethod auf die Bindung des 1. Arguments**

```python
def show_args(*args, method='method'):
    print(f'{method} bekommt {len(args)} Argument(e):')
    print('Argumente: ', end='')
    print(*args, sep=' und ')


class A:
    def f(*args):
        show_args(*args)

    @classmethod
    def g(*args):
        show_args(*args, method='method')

    @staticmethod
    def h(*args):
        show_args(*args, method='method')


a = A()
A, a  # Klasse und Instanz
```

```python
# A.f(1)
# a.f(1)
# A.g(1)
a.g(1)
# A.h(1)
# a.h(1)
```

**Zusätzlicher Konstruktor mit einer `classmethod`**

```python
class Person:
    @staticmethod
    def birthyear2age(birthyear):
        '''berechnet Alter aus Jahrgang'''
        return 2026 - birthyear

    @classmethod
    def from_birthyear(cls, name, birthyear):
        '''erstellt ein Objekt vom Typ Person'''
        return cls(name=name, age=cls.birthyear2age(birthyear))

    def __init__(self, name, age):
        '''initializiert eine Instanz der Klasse Person'''
        self.name = name
        self.age = age

    def birthday(self):
        '''inkrementiert self.age'''
        self.age = self.age + 1

    def __repr__(self):
        return f'Person(name={self.name!r}, age={self.age!r})'
```

```python
bob = Person.from_birthyear('bob', 2000)
bob
```

```python
Person.birthyear2age(2000) == bob.birthyear2age(2000) == 26
```

### Account, Rectangle und Circle Klassen

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposite(self, amount):
        self.balance += amount
        print(f'Einzahlung: {amount} CHF, new balance: {self.balance}')

    def withdraw(self, amount):
        if self.balance < amount:
            amount = self.balance
        self.balance -= amount
        print(f'Bezug: {amount} CHF, new balance: {self.balance}')

    def __repr__(self):
        return f'Account(owner={self.owner!r}, balance={self.balance!r})'
```

```python
account = Account('Bob', 100)
account
```

```python
account.deposite(20)
```

```python
account.withdraw(200)
```

```python
from math import pi as PI


class Rectangle:
    def __init__(self, lower_left, width, height):
        self.lower_left = lower_left
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return self.width * 2 + self.height * 2

    def move(self, dx, dy):
        x0, y0 = self.lower_left
        self.lower_left = (x0 + dx, y0 + dy)

    def draw(self, canvas, clear=True, fill_style=None, stroke_style='black'):
        x0, y0 = self.lower_left

        canvas.save()

        if clear:
            canvas.clear()
        if fill_style:
            canvas.fill_style = fill_style
            canvas.fill_rect(x0, y0, self.width, self.height)
        if stroke_style:
            canvas.stroke_style = stroke_style
            canvas.stroke_rect(x0, y0, self.width, self.height)

        canvas.restore()

    def __repr__(self):
        return f'Rectangle(lower_left={self.lower_left}, width={self.width}, height={self.height})'


class Circle:
    def __init__(self, center, radius):
        self.center = center
        self.radius = radius

    def area(self):
        return PI*self.radius**2

    def perimeter(self):
        return 2*PI*self.radius

    def move(self, dx, dy):
        x0, y0 = self.center
        self.center = (x0 + dx, y0 + dy)

    def draw(self, canvas, clear=True, fill_style=None, stroke_style='black'):
        x0, y0 = self.center

        canvas.save()

        if clear:
            canvas.clear()
        if fill_style:
            canvas.fill_style = fill_style
            canvas.fill_circle(x0, y0, self.radius)
        if stroke_style:
            canvas.stroke_style = stroke_style
            canvas.stroke_circle(x0, y0, self.radius)

        canvas.restore()

    def __repr__(self):
        return f'Circle(center={self.center}, radius={self.radius})'
```

```python
import widget_helpers as W
from ipywidgets import HBox


canvas1 = W.get_canvas()
canvas2 = W.get_canvas()
HBox([canvas1, canvas2])
```

```python
center, width, height, radius = (0, 0), 40, 20, 30
rect = Rectangle(center, width, height)
circle = Circle(center, radius)
rect, circle
```

```python
rect.move(10, 10)
circle.move(5, 5)
rect.draw(canvas1, fill_style='blue', stroke_style='red')
circle.draw(canvas2, clear=False, fill_style='orange', stroke_style='red')
```
