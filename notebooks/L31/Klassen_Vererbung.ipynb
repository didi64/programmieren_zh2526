### Vererbung


Der Sinn und Zweck von Vererbung:

* **Code-Wiederverwendung**:
  Gemeinsamer Code (Attribute und Methoden) muss nicht für jede Klasse neu geschreiben werden. Die Unterklassen erben ihn automatisch.
* **Vermeidung von Redundanz (DRY  "Don't Repeat Yourself")**:  
  Wenn sich etwas am grundlegenden Verhalten ändert, musst nur der Code der Basisklasse angepassen werden.    
* **Erweiterbarkeit und Spezialisierung**:
   Bestehende Klassen können für spezielle Anwendungsfälle erweitern oder anpassen werden, ohne das    Original zu verändern.

**Terminologie**:
- **Basisklasse (Superklasse, Elternklasse)**: Die Klasse, von der geerbt wird.

- **Unterklasse (Subklasse, Kindklasse)**: Die neue Klasse, die den Code übernimmt und anpasst

**Methode der Elternklasse überschreiben**:  
Definiert die Kindklasse eine Methode, die bereits in der Elternklasse definiert ist, 
überschreibt dies die Methode der Elternklasse.

**Methode der Elternklasse aufrufen**:  
  - `super().methode_der_Elternklasse(...)` ruft  
    `Elterklasse(self, ...)` auf.  
    `self` ist dabei das `self` der aufrufenden Klasse.

````python
class Animal:  # Basisklasse (Elternklasse)
    def __init__(self, name):
        self.name = name

    def talk(self):
        print('Grr!')

    def __repr__(self):
        return f'{type(self).__name__}(name={self.name!r})'
````

````python
animal = Animal(name='Some Animal')
animal.talk()
animal
````

````python
class Dog(Animal):  # Unterklasse (Kindklasse) erbt von Animal
    def talk(self):  # ueberschreine Methode talk der Elternklasse Animal
        print('Wuff!')
````

````python
dog = Dog(name='Bello')
dog.talk()
dog
````

````python
class Animal:
    def __init__(self, name, sound='Grr!'):
        self.name = name
        self.sound = sound

    def talk(self):
        print(f'{self.sound}')

    def what_am_I(self):
        print(f'I am a {type(self).__name__}')  # type(self) ist Klasse von self

    def __repr__(self):
        return f'{type(self).__name__}(name={self.name!r}, sound={self.sound!r})'
````

````python
animal = Animal(name='Some Animal')
animal.talk()
animal.what_am_I()
animal
````

````python
class Dog(Animal):
    def __init__(self, name, sound='Wuff!'):
        # rufe __init__ der Elternklasse auf, Animal.__init__(self, name, sound)
        super().__init__(name, sound)
````

````python
dog = Dog(name='Bello')
dog.talk()
dog.what_am_I()
print(f'My name is {dog.name}')  # self.name wurde von super().__init__ gesetzt
dog
````

````python
class Dog(Animal):
    def __init__(self, name, owner, sound='Wuff!'):
        self.owner = owner
        super().__init__(name, sound)

    # talk Methode ueberschreiben
    def talk(self):
        print(5*self.sound)

    def __repr__(self):
        return f'Dog(name={self.name!r}, owner={self.owner!r}, sound={self.sound!r})'
````

````python
dog = Dog(name='Bello', owner='Bob')
dog.talk()
````

````python
class Dog(Animal):
    #  __init__ erweitern
    def __init__(self, name, owner, sound='Wuff!'):
        self.owner = owner
        super().__init__(name, sound)

    # talk Methode erweitern
    def talk(self):
        print('Listen ...:', end=' ')
        # rufte Animal.talk(self) auf
        super().talk()

    def __repr__(self):
        return f'Dog(name={self.name!r}, owner={self.owner!r}, sound={self.sound!r})'
````

````python
dog = Dog(name='Bello', owner='Bob')
dog.talk()
dog.what_am_I()
dog
````

### Aufgabe
Es sollen Klassen `Circle(anchor, width)` und `Square(anchor, width)` erstellt werden, mit einer Methode
```python
def draw(self, canvas, clear=True,  fill_style=None, stroke_style='black'):
    '''zeichnet sich auf canvas
       ist clear gleich True wird die canvas vor dem Zeichnen geloescht.
       ist stroke_style nicht None, wird der Umriss in der Farbe stroke_style gezeichnet
       ist fill_style nicht None, wird der der Umriss in der Farbe fill_style gefuellt
    '''
```
`anchor` ist das Zentrum oder die Ecke, `width` der Radius oder Breite.  
Circle und Square sollen von der Klasse Shape erben.
Überschreibe jeweils die Methoden `filler` und `stroker` so, dass
sich Circle und Square auf die canvas zeichnen. Teste auch die Methode `move`.

````python
class Shape:
    def __init__(self, anchor, width):
        self.anchor = anchor
        self.width = width

    def move(self, dx, dy):
        x, y = self.anchor
        self.anchor = (x+dx, y+dy)

    def stroker(self, canvas, center, width):
        print(f'stroke {type(self).__name__} on canvas with center/corner {center} and width/radius {width}')

    def filler(self, canvas, center, width):
        print(f'stroke {type(self).__name__} on canvas at center/corner {center} and width/radius {width}')

    def draw(self, canvas, clear=True,  fill_style=None, stroke_style='black'):
        canvas.save()

        if clear:
            canvas.clear()
        if fill_style:
            canvas.fill_style = fill_style
            self.filler(canvas, self.anchor, self.width)
        if stroke_style:
            canvas.stroke_style = stroke_style
            self.stroker(canvas, self.anchor, self.width)

        canvas.restore()


class Circle(Shape):
    def __init__(self, anchor, width):
        super().__init__(anchor, width)


class Square(Shape):
    def __init__(self, anchor, width):
        super().__init__(anchor, width)
````

````python
import widget_helpers as W
from ipywidgets import HBox


canvas1 = W.get_canvas()
canvas2 = W.get_canvas()
HBox([canvas1, canvas2])
````

````python
circle = Circle((50, 50), 30)
square = Square((10, 10), 60)
````

````python
circle.draw(canvas1, fill_style='red', stroke_style='blue')
square.draw(canvas2, fill_style='red', stroke_style='blue')
````
