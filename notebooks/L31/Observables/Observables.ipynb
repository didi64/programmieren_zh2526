### Observables

Wir nennen Objekt `obj` observable, wenn es eine Methode  
> `register_callback(f)`

zur Verfügung stellt, die es erlaubt, eine Funktion `f(event, data)` als
Callback zu registrieren, welche dann beim Eintreffen eines Events aufgerufen wird.



```python
import widget_helpers as W
from ipywidgets import Select
from IPython.display import display

out = W.get_out()
select = Select(options=('A', 'B'), value=None, rows=2)


@out.capture(clear_output=True)
def on_change(change):
    print(f'Your choice: {change.new}')


select.observe(on_change, names='value')
display(select, out)
```

```python
class Game_1:
    initial_stacks = [(2, 1, 0), (), ()]

    def __init__(self):
        self.stacks = []
        self.success = None

    def update(self, event, data=None):
        print(f'event: {event}, data: {data}')

    def new_game(self):
        self.success = False
        self.stacks = [list(stack) for stack in self.initial_stacks]
        self.update('new_game')

    def move(self, i, j):
        if not (0 <= i <= 2 and 0 <= j <= 2 and i != j):
            return

        if not self.stacks[i] or (self.stacks[j] and self.stacks[j] < self.stacks[i]):
            return

        self.stacks[j].append(self.stacks[i].pop())
        if j != 0 and len(self.stacks[j]) == len(self.initial_stacks[0]):
            self.success = True

        self.update('move', data=(i, j))

    def __repr__(self):
        return f'stacks: {self.stacks}, success: {self.success}'
```

```python
from my_observable import Observable, notify


class Game(Observable):
    initial_stacks = [(2, 1, 0), (), ()]

    def __init__(self):
        self.stacks = []

    @notify
    def new_game(self):
        self.success = False
        self.stacks = [list(stack) for stack in self.initial_stacks]

    @notify
    def move(self, i, j):
        if not (0 <= i <= 2 and 0 <= j <= 2 and i != j):
            return

        if not self.stacks[i] or (self.stacks[j] and self.stacks[j] < self.stacks[i]):
            return

        self.stacks[j].append(self.stacks[i].pop())
        if j != 0 and len(self.stacks[j]) == len(self.initial_stacks[0]):
            self.success = True

        return (i, j)

    def __repr__(self):
        return f'stacks: {self.stacks}, success: {self.success}'
```

```python
game = Game()
game.register_callback(lambda event, data: print(f'Callback: {event}, {data}'))
```
