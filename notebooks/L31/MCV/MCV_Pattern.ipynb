### Das Model-View-Controller Pattern (MCV) am Beispiel eines Spiels
 
 1. **Das Model** (Game)
* Die Klasse Game verwaltet den Zustand des Spiels und die Spiellogik.
* Das Model ist komplett unabhängig von der Darstellung (View). Es weiß nicht, wie das Spiel gezeichnet wird (ob mit ipycanvas, in der Konsole oder 3D). Es signalisiert Änderungen durch AUfruf einer Callback-Funktion (z.B. `update(event, date)`).

2. **Die View**
* Die Klasse View kapselt das Canvas-Objekt und ist ausschließlich für die grafische Repräsentation zuständig.
* Die View beobachtet das Model, indem es ein Callback beim Game registriert (bisher: View überschreibt `game.update`).
* Sobald das Model via Callback `update(event, data)` signalisiert, dass sich etwas Relevantes geändert hat,
  analysiert `update(event, data)` die Argumente und passt die Darstellung an. Dazu kann die View auch auf das Game-Objekt zugreifen.

3. **Der Controller**
* Der Controller benutzt das Canvas-Objekt der View um auf Tasten-und Mausevents zu hören.
* Der Controller nimmt die Eingaben des Benutzers entgegen und übersetzt diese in Befehle für das Model (die Methoden von Game).

**Der Datenfluss**:   
1. User-Aktion: Der Spieler drückt eine Taste -> Das Canvas-Objekt der View fängt das Event ab.
2. Controller greift ein: Da der Controller die Event-Listener (`on_key_down` etc.) auf dem Canvas registriert hat, verarbeitet er die Eingabe.
3. Model-Aktualisierung: Der Controller konsultiert den Callbacks-Dict und ruft die entsprechende Methode im Game (Model) auf.
4. Zustandsänderung: Das Game verändert seinen internen Zustand (z. B. Spielerposition).
5. Benachrichtigung: Das Game benachrichtigt die View durch Aufruf des von der View registrierten Callbacks `update`
6. View-Aktualisierung: Die View malt mit `view.update(event, data)` den aktuellen Spielzustand auf die Canvas.
